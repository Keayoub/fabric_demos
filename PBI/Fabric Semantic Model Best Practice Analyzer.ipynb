# Semantic Model Best Practice Analyzer
Companion to `Fabric Read-Only Semantic Model Diagnostics.ipynb`. This locally authored notebook calls Microsoft's Semantic Link Labs BPA API; it is not a copy of an official Microsoft notebook.

**Purpose:** Identify model design, DAX, and performance rule violations. Review relationship-related findings first for your scenario. Findings are recommendations, not proof of incorrect totals; this does not repair the model or analyze report visual timings.

**Requirements:** Microsoft Fabric notebook, `semantic-link-labs`, an authorized target workspace/model, and working capacity/XMLA access. TOM-based dependencies can require elevated model permissions even for analysis. No refresh, model edit, or lakehouse export is requested. Standard scans still consume compute; extended scans fetch VertiPaq statistics and may be more expensive. Notebook outputs can contain sensitive metadata.

**Setup:** If `sempy_labs` is unavailable, run `%pip install semantic-link-labs` in a separate Fabric cell before importing, or use an approved Fabric Environment with the package installed. Use your organization's approved package version. Installation changes the notebook environment, not the semantic model.

[Microsoft example](https://github.com/microsoft/semantic-link-labs/wiki/Code-Examples#model-best-practice-analyzer) | [API reference](https://semantic-link-labs.readthedocs.io/en/stable/sempy_labs.html#sempy_labs.run_model_bpa)


```python
WORKSPACE_NAME = ""
MODEL_NAME = ""
RUN_EXTENDED_BPA = False

if not WORKSPACE_NAME.strip() or not MODEL_NAME.strip():
    raise ValueError("Set WORKSPACE_NAME and MODEL_NAME to the exact Fabric workspace and semantic model names.")
```

```python
import sempy_labs as labs

print(f"Running BPA for {MODEL_NAME!r} in {WORKSPACE_NAME!r}; extended={RUN_EXTENDED_BPA}; export=False.")
labs.run_model_bpa(
    dataset=MODEL_NAME,
    workspace=WORKSPACE_NAME,
    extended=RUN_EXTENDED_BPA,
    export=False,
)
```

## Interpret the BPA output
Review rule descriptions, affected objects, and severity in the rendered BPA results. Cross-reference relationship endpoints in the inventory notebook; check cardinality, filter direction, active status, and intended business behavior before making any change. Not every rule applies to every model.

An empty/clean scan means no violations of the evaluated rules, not that every relationship or measure is correct. Missing dimension keys and incorrect filter context need separate checks. To include advanced rules, set `RUN_EXTENDED_BPA=True` and rerun after approving the additional workload.

Errors are intentionally not suppressed: resolve package/API-version, permissions, model identity, capacity, or XMLA errors before trusting results. No live scan was run when this notebook was created. Changes and repairs must be reviewed and performed separately.
